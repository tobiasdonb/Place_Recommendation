```python
import requests
import datetime 
import os 
```

Ini notebook untuk saya mencoba2 menggunakan API mengenai maps dari google

Saya ingin mencoba menggunakan API Distance matrix dari google untuk menghitung waktu perjalanan dari titik A ke titik B di python

```python
#sebelumnya saya sudah mendeklarasikan variabel berisi API key di os demi keamanan code
api_key = os.getenv("MAPS_API_KEY")
if api_key is None:
    raise ValueError("API key not found")
```

```python
#function waktu tempuh titik a ke b 
def get_travel_time(origin,destination,departure_time):
    
    #parameter yang digunakan agar API bisa mengetahui mode, titik awal,destinasi dan waktu keberangkatan
    params = {
        "origins": origin,
        "destinations":destination,
        "mode":"driving",
        "departure_time":departure_time,
        "key":api_key
    }
    res = requests.get("https://maps.googleapis.com/maps/api/distancematrix/json",params=params)
    data = res.json()
    #memasukan data yang kita punya ke masing masing 
    destination_addresses = data["destination_addresses"]
    origin_addresses = data["origin_addresses"]
    distance = data["rows"][0]["elements"][0]["distance"]["value"]
    duration_in_second = data["rows"][0]["elements"][0]["duration_in_traffic"]["value"]
    duration_in_minute = str(duration_in_second //60) + ' minute ' + str(duration_in_second %60) + ' seconds '

    return {
        "origin_addresses":origin_addresses,
        "destination_addresses":destination_addresses,
        "distance in meters":distance,
        "duration in second":duration_in_second,
        "duration in minute":duration_in_minute,
    }
    






```

```python
#isi python dict a.k.a hasil parsingan dari json (kalau kita return variabel data di dalam function)
get_travel_time('MS FIT','Fitnespluss Sanur','now')
```
